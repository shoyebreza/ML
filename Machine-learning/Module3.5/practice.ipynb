## A2. Manual Min-Max on a tiny set

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
scores = [20, 25, 30, 50]

# Step 1: Find the minimum and maximum values in the dataset
min_score = min(scores)
max_score = max(scores)

print(f"Original scores: {scores}")
print(f"Minimum score: {min_score}")
print(f"Maximum score: {max_score}")

# Step 2: Apply the Min-Max scaling formula for each score
# Formula: scaled_value = (original_value - min_value) / (max_value - min_value)

scaled_scores = []
print("\nScaling steps:")
for score in scores:
    scaled_value = (score - min_score) / (max_score - min_score)
    scaled_scores.append(scaled_value)
    print(f"  Score {score}: ({score} - {min_score}) / ({max_score} - {min_score}) = {scaled_value:.4f}")

print(f"\nScaled scores: {scaled_scores}")
```

## A3. Z-scores on a subset

```python
x = [8, 9, 11]

print(f"Original data: {x}")

# Step 1: Compute the mean
mean_x = sum(x) / len(x)
print(f"\nMean of x: {mean_x:.2f}")

# Step 2: Compute the population standard deviation
# Calculate the variance first
variance_x = sum([(val - mean_x)**2 for val in x]) / len(x)
std_dev_x = variance_x**0.5
print(f"Population Standard Deviation of x: {std_dev_x:.2f}")

# Step 3: Standardize each value (Z-score)
# Formula: z = (value - mean) / standard_deviation
standardized_x = []
print("\nStandardization steps:")
for val in x:
    z_score = (val - mean_x) / std_dev_x
    standardized_x.append(z_score)
    print(f"  Value {val}: ({val} - {mean_x:.2f}) / {std_dev_x:.2f} = {z_score:.4f}")

print(f"\nStandardized x (Z-scores): {standardized_x}")
```

## A4. Robust scaling ingredients

```python
import numpy as np

y = [5, 6, 6, 7, 50]
print(f"Original data: {y}")

# Calculate median (Q2)
median_y = np.median(y)

# Calculate Q1 (25th percentile)
Q1_y = np.percentile(y, 25)

# Calculate Q3 (75th percentile)
Q3_y = np.percentile(y, 75)

# Calculate IQR
IQR_y = Q3_y - Q1_y

print(f"Median (Q2): {median_y}")
print(f"Q1: {Q1_y}")
print(f"Q3: {Q3_y}")
print(f"IQR: {IQR_y}")
```

## A5. Nominal or ordinal

```python
import numpy as np

# Data y and pre-calculated values (from cell rxyBlseNKzeo)
y = [5, 6, 6, 7, 50]
Q1_y = np.percentile(y, 25)
Q3_y = np.percentile(y, 75)
IQR_y = Q3_y - Q1_y

print(f"Original data: {y}")
print(f"Q1: {Q1_y}")
print(f"Q3: {Q3_y}")
print(f"IQR: {IQR_y}")

# Calculate the lower and upper bounds for outliers
lower_bound = Q1_y - 1.5 * IQR_y
upper_bound = Q3_y + 1.5 * IQR_y

print(f"\nLower bound for outliers: {lower_bound}")
print(f"Upper bound for outliers: {upper_bound}")

# Identify outliers
outliers = [val for val in y if val < lower_bound or val > upper_bound]

if outliers:
    print(f"\nIdentified outliers: {outliers}")
else:
    print("\nNo outliers detected.")
```

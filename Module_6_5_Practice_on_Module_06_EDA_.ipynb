## Section 4 — Identify Target Column and Feature Types

Separate the target and features into numeric and categorical groups.


```python
# TODO:
# 1. Identify your target column name
target = 'diagnosed_diabetes'

# 2. Create a list of numeric feature columns
numeric_features = df.select_dtypes(include=[np.number]).columns.tolist()
numeric_features.remove(target)

# 3. Create a list of categorical feature columns
categorical_features = df.select_dtypes(include=['object']).columns.tolist()

# Identify binary numerical columns that are often treated as categorical
binary_categorical_columns = ['family_history_diabetes', 'hypertension_history', 'cardiovascular_history']

# Add these to categorical_features and remove from numeric_features
for col in binary_categorical_columns:
    if col in numeric_features:
        numeric_features.remove(col)
        categorical_features.append(col)

# 4. Print all three
print('Target: ', target)
print('Numeric features: ', numeric_features)
print('Categorical features: ', categorical_features)
```

### Student Notes and Observations

_Which columns did you treat as numeric? Which are categorical? Any columns that were hard to classify?_

## Section 10 — Pairplot and Correlation Analysis

Explore relationships between features and correlations.

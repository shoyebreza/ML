# A3. Z-scores on a subset


````python
x = [8, 9, 11]

print(f"Original data: {x}")

# Step 1: Compute the mean
mean_x = sum(x) / len(x)
print(f"\nMean of x: {mean_x:.2f}")

# Step 2: Compute the population standard deviation
# Calculate the variance first
variance_x = sum([(val - mean_x)**2 for val in x]) / len(x)
std_dev_x = variance_x**0.5
print(f"Population Standard Deviation of x: {std_dev_x:.2f}")

# Step 3: Standardize each value (Z-score)
# Formula: z = (value - mean) / standard_deviation
standardized_x = []
print("\nStandardization steps:")
for val in x:
    z_score = (val - mean_x) / std_dev_x
    standardized_x.append(z_score)
    print(f"  Value {val}: ({val} - {mean_x:.2f}) / {std_dev_x:.2f} = {z_score:.4f}")

print(f"\nStandardized x (Z-scores): {standardized_x}")
````

## B5. Vectors and alignment

````python
import numpy as np

# Define the vectors
a = np.array([3, -1, 2])
b = np.array([4, 0, -2])
c = np.array([-6, 2, -4])

print(f"Vector a: {a}")
print(f"Vector b: {b}")
print(f"Vector c: {c}\n")

# a) Compute a·b and a·c
dot_ab = np.dot(a, b)
dot_ac = np.dot(a, c)

print(f"a·b = {dot_ab}")
print(f"a·c = {dot_ac}\n")

# b) Compare signs and magnitudes to comment on the alignment
print("Alignment Analysis:")

# Alignment of a with b
if dot_ab > 0:
    print(f"- a and b have a positive dot product ({dot_ab}), indicating they are generally aligned (angle < 90 degrees).")
elif dot_ab < 0:
    print(f"- a and b have a negative dot product ({dot_ab}), indicating they are generally anti-aligned (angle > 90 degrees).")
else:
    print(f"- a and b have a dot product of zero, indicating they are orthogonal (perpendicular).")

# Alignment of a with c
if dot_ac > 0:
    print(f"- a and c have a positive dot product ({dot_ac}), indicating they are generally aligned (angle < 90 degrees).")
elif dot_ac < 0:
    print(f"- a and c have a negative dot product ({dot_ac}), indicating they are generally anti-aligned (angle > 90 degrees).")
else:
    print(f"- a and c have a dot product of zero, indicating they are orthogonal (perpendicular).")

print(f"  Since a·c ({dot_ac}) is exactly -2 times a·b ({dot_ab}), it suggests c is exactly anti-parallel to a with a proportional magnitude relative to b.\n")

# c) L2 normalize a and give the normalized vector to three decimals
# Calculate the L2 norm (Euclidean norm)
norm_a = np.linalg.norm(a)
normalized_a = a / norm_a

print(f"L2 Norm of a: {norm_a:.4f}")
print(f"Normalized vector a (to three decimals): {np.round(normalized_a, 3)}")
````

## B6. Two distances, different vibes

````python
import numpy as np
from scipy.spatial.distance import euclidean, cityblock # cityblock is Manhattan distance

# Define the points
P1 = np.array([2, 3])
P2 = np.array([5, 7])
P3 = np.array([2, 10])

print(f"P1: {P1}, P2: {P2}, P3: {P3}\n")

# a) Compute Euclidean and Manhattan distances for all pairs

# P1 and P2
d_euclidean_p1p2 = euclidean(P1, P2)
d_manhattan_p1p2 = cityblock(P1, P2)
print(f"Distances between P1 and P2:")
print(f"  Euclidean: {d_euclidean_p1p2:.4f}")
print(f"  Manhattan: {d_manhattan_p1p2:.4f}\n")

# P1 and P3
d_euclidean_p1p3 = euclidean(P1, P3)
d_manhattan_p1p3 = cityblock(P1, P3)
print(f"Distances between P1 and P3:")
print(f"  Euclidean: {d_euclidean_p1p3:.4f}")
print(f"  Manhattan: {d_manhattan_p1p3:.4f}\n")

# P2 and P3
d_euclidean_p2p3 = euclidean(P2, P3)
d_manhattan_p2p3 = cityblock(P2, P3)
print(f"Distances between P2 and P3:")
print(f"  Euclidean: {d_euclidean_p2p3:.4f}")
print(f"  Manhattan: {d_manhattan_p2p3:.4f}\n")

# b) Which distance is more sensitive to a single large jump in one coordinate
print("b) Sensitivity to a single large jump:")
print("   Manhattan distance is more sensitive to a single large jump in one coordinate because it sums the absolute differences across dimensions, while Euclidean distance squares these differences, diminishing the relative impact of a single large jump compared to multiple smaller changes.")

# c) Scale y by 10 and recompute d(P1, P2) for both distances, then explain the effect in one line

P1_scaled = np.array([P1[0], P1[1] * 10])
P2_scaled = np.array([P2[0], P2[1] * 10])

print(f"\nScaled P1: {P1_scaled}, Scaled P2: {P2_scaled}\n")

d_euclidean_p1p2_scaled = euclidean(P1_scaled, P2_scaled)
d_manhattan_p1p2_scaled = cityblock(P1_scaled, P2_scaled)

print(f"Distances between Scaled P1 and Scaled P2:")
print(f"  Euclidean: {d_euclidean_p1p2_scaled:.4f}")
print(f"  Manhattan: {d_manhattan_p1p2_scaled:.4f}\n")

print("c) Effect of scaling on distances:")
print("   Scaling a coordinate by 10 increases both Euclidean and Manhattan distances by a factor that depends on the magnitude of the scaled coordinate's difference, effectively amplifying its contribution to the overall distance.")
````

### Subtask:
Combine C-Data-1 and C-Data-2, then use sklearn's ColumnTransformer to apply one-hot encoding to nominal features, ordinal encoding to ordinal features (with specified mappings), and appropriate scaling (Min-Max, Standard, Robust) to numerical features. Display the shape and first few rows of the transformed data.


**Reasoning**:
The first step is to import all the necessary libraries as specified in the instructions to prepare for data merging and transformation.



````python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, MinMaxScaler, StandardScaler, RobustScaler

print("Required libraries imported successfully.")
````

**Reasoning**:
Now that the libraries are imported, I will merge C_Data_1 and C_Data_2 into a single DataFrame `combined_df` using the 'ID' column, and then drop the 'ID' column as it's not a feature for transformation.



````python
combined_df = pd.merge(C_Data_1, C_Data_2, on='ID', how='inner')
combined_df = combined_df.drop('ID', axis=1)

print("Combined DataFrame created and 'ID' column dropped. Shape:", combined_df.shape)
print("First 5 rows of combined_df:\n", combined_df.head())
````

**Reasoning**:
Now that the `combined_df` is prepared, I will define the column lists for nominal, ordinal, Min-Max, Standard, and Robust scaling, along with the specific category orderings for ordinal features, as outlined in the subtask instructions.



````python
nominal_features = ['Internet', 'City']
ordinal_features = ['Education', 'Satisfaction', 'Rating']
minmax_features = ['Age', 'GPA']
standard_features = ['Hours_Study', 'Temp_C']
robust_features = ['Income_BDT', 'Transactions']

education_categories = ['High School', 'Bachelor', 'Master', 'PhD']
satisfaction_categories = ['Medium', 'High', 'Very High']
rating_categories = [3, 4, 5]

print("Feature lists and ordinal categories defined.")
````

**Reasoning**:
With the feature lists and ordinal categories defined, I will now create the `ColumnTransformer` to apply all specified preprocessing steps (one-hot encoding, ordinal encoding, Min-Max scaling, Standard scaling, and Robust scaling) to the `combined_df`. Then, I will fit and transform the data and display the shape and first five rows of the resulting transformed data as per the instructions.



````python
preprocessor = ColumnTransformer(
    transformers=[
        ('onehot', OneHotEncoder(handle_unknown='ignore'), nominal_features),
        ('ordinal', OrdinalEncoder(categories=[education_categories, satisfaction_categories, rating_categories]), ordinal_features),
        ('minmax', MinMaxScaler(), minmax_features),
        ('standard', StandardScaler(), standard_features),
        ('robust', RobustScaler(), robust_features)
    ])

transformed_data = preprocessor.fit_transform(combined_df)

print("Shape of transformed_data:", transformed_data.shape)
print("\nFirst 5 rows of transformed_data:\n", transformed_data[:5])
````
